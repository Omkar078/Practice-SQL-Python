```python
##Problem Statement:
##Given a list of product names, write a program that:

###Sorts them alphabetically (A → Z)
###Sorts them in reverse order (Z → A)
### Finds the product that comes first alphabetically without sorting the original list
```

```python
###Sorts them alphabetically (A → Z)
products = ["Wireless Mouse", "Apple Watch", "Zoom Lens", "Mechanical Keyboard", "USB Hub", "Bluetooth Speaker"]

p = sorted(products)

print("Original List:", p)

###nSorts them in reverse order (Z → A)

q = sorted(products, reverse = True)

print("Sorted A - Z:", q)

### Finds the product that comes first alphabetically without sorting the original list
r = min(products)

print(r)

```

```python

##Problem Statement:
##Given a list of order IDs, write a program that:

##Finds all duplicate order IDs
##Returns a cleaned list with duplicates removed, preserving original order
##Prints count of duplicates found
```

```python
order_ids = ["ORD-101", "ORD-205", "ORD-101", "ORD-309", "ORD-205", "ORD-412", "ORD-309", "ORD-515"]

##Finds all duplicate order IDs
a = set(order_ids)

print(a)

##Returns a cleaned list with duplicates removed, preserving original order
cleaned_list = []

for order in order_ids:
    if order not in cleaned_list:
        cleaned_list.append(order)

print(cleaned_list)

##Prints count of duplicates found
duplicates = []

for order in cleaned_list:
    if order_ids.count(order) > 1 and order not in duplicates:
        duplicates.append(order)

print(len(duplicates))
```

```python
###Customer Transaction Frequency Ranker
### A fraud analytics team wants to rank customers by how frequently they transact. They need the top 3 most active customers and want the transactions sorted by amount within each customer's record.
```

```python
transactions = [
    ("Riya Shah", 4500), ("Amit Verma", 1200), ("Riya Shah", 8900),
    ("Priya Nair", 300),  ("Amit Verma", 5600), ("Riya Shah", 2100),
    ("Amit Verma", 4400), ("Neha Joshi", 750),  ("Priya Nair", 1800),
    ("Amit Verma", 900),  ("Riya Shah", 6700),  ("Neha Joshi", 2200)
]
```

```python
# Groups transactions by customer name using a dictionary

a = {}

for name, amnt in transactions:
    if name not in a:
        a[name] = []

    a[name].append(amnt)

print(a)
```

```python
### Counts each customer's transaction frequency

a = {}

for name, amnt in transactions:
    if name not in a:
        a[name] = 0

    a[name] += 1

print(a)


```

```python
### Sorts customers by frequency (highest first)

a =  {}

for name, amnt in transactions:
    if name not in a:
        a[name] = 0

    a[name] += 1

sorted_customers = sorted(a.items(), key= lambda x: x[1], reverse= True)
print(sorted_customers)
```

```python
### Within each customer, sorts their transaction amounts in descending order

a = {}

for name, amnt in transactions:
    if name not in a:
        a[name] =  []

    a[name].append(amnt)

for name in a:
    a[name].sort(reverse = True)

print(a)

```

```python
transactions = [
    ("Alice", 120),
    ("Bob", 75),
    ("Alice", 200),
    ("Bob", 50),
    ("Alice", 90),
    ("Charlie", 300),
    ("Charlie", 150),
    ("Charlie", 100),
    ("David", 500)
]

# Store transactions customer-wise
customer_transactions = {}

for name, amount in transactions:
    if name not in customer_transactions:
        customer_transactions[name] = []

    customer_transactions[name].append(amount)

# Sort each customer's transactions in descending order
for name in customer_transactions:
    customer_transactions[name].sort(reverse=True)

# Sort customers by number of transactions (highest first)
top_customers = sorted(
    customer_transactions.items(),
    key=lambda x: len(x[1]),
    reverse=True
)

# Print top 3 most active customers
for name, amounts in top_customers[:3]:
    total_spend = sum(amounts)

    print(f"Customer: {name}")
    print(f"Transactions: {amounts}")
    print(f"Total Spend: {total_spend}")
    print()
```
